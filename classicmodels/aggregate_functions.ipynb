How many products are ordered for every order.

```python
q1 = "SELECT orderNumber, SUM(quantityOrdered) AS quantity_of_products FROM orderDetails GROUP BY orderNumber"
```

Total Price for every order.

```python
q2 = """SELECT orderNumber, sum(CAST(priceEach as DECIMAL(10, 2)) * quantityOrdered) AS total_Price_for_Order From orderDetails 
GROUP BY orderNumber;"""
```

How many products are in every productline

```python
q3 = "SELECT productLine, SUM(quantityinstock) AS total_quantity_In_Stock FROM products GROUP BY productLine"
```

Average manufacturer's suggested retail price for every productline

```python
q4 = "SELECT productLine, AVG(msrp) AS average_msrp FROM products GROUP BY productLine"
```

Number of customers by country

```python
q5 = "SELECT country, count(customerNumber) AS number_of_customers FROM customers GROUP BY country"
```

Minimum and maximum credit limit by country

```python
q6 = "SELECT country, MIN(creditLimit) FROM customers GROUP BY country"
q7 = "SELECT country, MAX(creditLimit) FROM customers GROUP BY country"
```

Connection with python

```python
import psycopg2

cnx = psycopg2.connect(
    host="localhost",
    database="classicmodels",
    user="postgres",
    password=1234
)

cursor = cnx.cursor()
```

```python
print("Amount of products ordered for every order.\n")

cursor.execute(q1)

for row in cursor.fetchall():
    print("Order Number:", row[0], "Amount of Products:", row[1])  
```

```python
print("Total Price for every order.\n")

cursor.execute(q2)

for row in cursor.fetchall():
    print("Order Number:", row[0], "Total Price:", row[1])
```

```python
print("Amount of products in every productline\n")

cursor.execute(q3)

for row in cursor.fetchall():
    print("Product Line:", row[0], "\nTotal Quantity in Stock:", row[1])
    print("----------------------------------------------------")
```

```python
print("Average manufacturer's suggested retail price for every productline\n")

cursor.execute(q4)

for row in cursor.fetchall():
    print("Product Line:", row[0], "\nAverage MSRP:", row[1])
    print("----------------------------------------------------")
```

```python
print("Number of customers for every country\n")
cursor.execute(q5)

for row in cursor.fetchall():
    print("Country:", row[0], "\nNumber of Customers:", row[1])
    print("----------------------------------------------------")
```

```python
print("Minimum Credit Limit for every country\n")

cursor.execute(q6)

for row in cursor.fetchall():
    print("Country:", row[0], "\nMinimum Credit Limit:", row[1])
    print("----------------------------------------------------")
```

```python
print("Maximum Credit Limit for every country\n")

cursor.execute(q7)

for row in cursor.fetchall():
    print("Country:", row[0], "\nMaximum Credit Limit:", row[1])
    print("----------------------------------------------------")
```
