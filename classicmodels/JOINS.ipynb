Join `products` and `productlines` tables and display all information from both tables.

```python
SELECT * FROM products  JOIN productLines ON products.productline = productLines.productline;
```

Join `customers` and `employees` tables and display all information from both tables.

```python
SELECT  customerNumber, employeeNumber FROM customers right JOIN employees ON customers.salesrepemployeenumber = employees.employeenumber
```

Companies that do not have an assigned employee

```python
SELECT  customerNumber, employees.employeenumber FROM customers LEFT JOIN employees ON customers.salesrepemployeenumber = employees.employeenumber WHERE employees.employeenumber is null 
```

Employees without an assingned company.

```python
SELECT  employeeNumber, customerNumber FROM customers RIGHT JOIN employees ON customers.salesrepemployeenumber = employees.employeenumber WHERE customers.customerNumber is null
```

Eployees who have more than one company assigned to them.

```python
SELECT  employees.employeenumber, firstName, count(customers.customerNumber) AS nr_of_costumers FROM customers JOIN employees ON customers.salesrepemployeenumber = employees.employeenumber group by  employees.employeenumber, firstName HAVING COUNT(customers.customerNumber) > 1;
```

SQL query that will return the following information from the database:

-  order ID (orders table),
- order status (orders table),
- code of the ordered product (orderdetails table),`
- quantity of the ordered product (orderdetails table),
- name of the ordered product (products table).

```python
SELECT orders.orderNumber, orders.status, orderDetails.productCode, products.productName, orderDetails.quantityordered FROM orders 
LEFT JOIN orderDetails on orders.orderNumber = orderDetails.orderNumber
LEFT JOIN products on orderDetails.productCode = products.productCode
```
