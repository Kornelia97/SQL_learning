Average buy price

```python
import psycopg2

cnx = psycopg2.connect(
host = 'localhost',
user = 'postgres',
password = 1234,
dbname = 'classicmodels'
)

cursor= cnx.cursor()
```

```python
cursor.execute("SELECT buyPrice FROM products")

sum_buy = 0
row_count = cursor.rowcount # counts rows in cursor object
for row in cursor:
    sum_buy += row[0]
avg_buy = sum_buy / row_count
print(avg_buy)
```

Minimum and maximum Manufacturers Suggested Retail Price

```python
import math



max_price = 0
min_price = math.inf #infinity

cursor= cnx.cursor()
cursor.execute("SELECT msrp  FROM products") # msrp = Manufacturers Suggested Retail Price

for row in cursor:
    if row[0] < min_price:
        min_price = row[0]
    if row[0] > max_price:
        max_price = row[0]
print("Minimum Manufacturers Suggested Retail Price: ", min_price)
print("Maximum Manufacturers Suggested Retail Price: ", max_price)
    
```
