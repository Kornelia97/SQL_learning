```python
import psycopg2


connection = psycopg2.connect(
    host = 'localhost',
    user = 'postgres',
    password = 1234,
    dbname = 'classicmodels'
)

cursor = connection.cursor()
```

1. Change the work location of employee Larry Bott from the office with ID 7 to the office with ID 5.

```python
cursor.execute("UPDATE employees SET officecode = 5 WHERE firstName = 'Larry'")
```

2. Change creditlimit for company 'Tekni Collectables Inc.' to 60 000.

```python
cursor.execute("UPDATE customers SET creditLimit = 60000 WHERE customerName = 'Tekni Collectables Inc.' ")
```

3. Change the number of available units for all products belonging to the Trucks and Buses product line to 0.

```python
cursor.execute("UPDATE products SET quantityInStock = 0 WHERE productLine = 'Trucks and Buses'")
```

4. Change the quantity of items with code S18_2957 in order number 10103 to 60.
